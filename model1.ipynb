```python
import glob
import os

import random
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

import torch
import torch.nn as nn
from torch.utils.data import DataLoader

import src.loss as loss
import Models
from src.dataset import Dataset

# import shutil
# if not os.path.isdir(DATASET_PATH):
#     raise FileNotFoundError(
#         f"There is no folder at {DATASET_PATH}.\n"
#         "Copy the real path from the Input panel on the right: expand the "
#         "competition data, hover over its folder, and click the copy-path icon."
#     )

# shutil.copy(f"{DATASET_PATH}/metric.py", "metric.py")
# from metric import rle_encode, rle_decode, global_dice, write_submission

from src.metric import global_dice, write_submission
```

SETTINGS

```python
SESSION_ID = "model01_dice_bce_loss"

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Paths
DATASET_PATH = "/home/dinis/Desktop/Universidade/Cadeiras/Ano4-Q1/AE4353 Artificial Intelligence for Aerospace Control and Operations/Competition/Data/ae4353-2026"

OUTPUT_BASE_PATH = "/home/dinis/Desktop/Universidade/Cadeiras/Ano4-Q1/AE4353 Artificial Intelligence for Aerospace Control and Operations/Competition/Output"

OUTPUT_PATH = os.path.join(OUTPUT_BASE_PATH, SESSION_ID)

# Dataset
BATCH_SIZE = 8
NUM_WORKERS = 4
OVERFIT_MODE = False
OVERFIT_NUM_IMAGES = 50
VAL_SPLIT = 0.2

# Model and Training
MAX_EPOCHS = 3
THRESHOLD_RANGE = np.arange(0.1, 0.9, 0.05)
LEARNING_RATE = 1e-3

model = Models.model01.Contrail_CNN()
criterion = loss.DiceBCELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE)

# Plot
PLOT_INPUT = False
PLOT_OUTPUT = True

# Seed
SEED = 2006
```

INITIALIZATION

```python
os.makedirs(OUTPUT_PATH, exist_ok=True)

train_files = sorted(glob.glob(f"{DATASET_PATH}/train_*.h5"))
test_files = sorted(glob.glob(f"{DATASET_PATH}/test_*.h5"))

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)                     # seeds the GPU as well
torch.backends.cudnn.benchmark = False      # the same algorithms on every run
torch.backends.cudnn.deterministic = True

def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)
```

DATASET LOADING

```python
train_dataset = Dataset(train_files, is_train=True)

if OVERFIT_MODE:
    train_dataset = torch.utils.data.Subset(train_dataset, range(50))


val_size = int(len(train_dataset) * VAL_SPLIT)
train_size = len(train_dataset) - val_size
train_dataset, val_dataset = torch.utils.data.random_split(train_dataset, [train_size, val_size], generator=torch.Generator().manual_seed(SEED))

test_dataset = Dataset(test_files, is_train=False)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    num_workers=NUM_WORKERS,
    worker_init_fn=seed_worker,
    generator=torch.Generator().manual_seed(SEED),
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    num_workers=NUM_WORKERS,
    worker_init_fn=seed_worker,
    generator=torch.Generator().manual_seed(SEED),
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    num_workers=NUM_WORKERS,
    worker_init_fn=seed_worker,
    generator=torch.Generator().manual_seed(SEED),
    shuffle=True
)

if PLOT_INPUT:

    positive_indices = []
    for index in range(len(train_dataset)):
        _, mask = train_dataset[index]
        if torch.any(mask > 0):
            positive_indices.append(index)

    print(f"Found {len(positive_indices)} images containing contrails.")

    n_to_plot = min(6, len(positive_indices))

    selected_indices = positive_indices[:n_to_plot]

    fig, axes = plt.subplots(n_to_plot, 2, figsize=(10, 4 * n_to_plot))
    axes = np.atleast_2d(axes)

    for row, index in enumerate(selected_indices):
        image, mask = train_dataset[index]
        image_np = image.permute(1, 2, 0).numpy()
        mask_np = mask.squeeze(0).numpy()

        axes[row, 0].imshow(image_np)
        axes[row, 0].set_title(f"Image {index}")
        axes[row, 0].axis("off")

        axes[row, 1].imshow(image_np)
        axes[row, 1].imshow(mask_np, cmap="Reds", alpha=0.5)
        axes[row, 1].set_title("Contrail mask")
        axes[row, 1].axis("off")

    plt.tight_layout()
    plt.show()

```

TRAINING

```python
model.to(device)

loss_history = []
validation_history = {
    "best_dice": [],
    "best_threshold": [],
    "false_positives": [],
    "false_negatives": []
}

validation_all_metrics_history = []

best_global_dice = 0.0
best_threshold = 0.0

for epoch in range(MAX_EPOCHS):
    
    # Training

    model.train()
    running_loss = 0.0

    for images, masks in train_loader:
        images = images.to(device)
        masks = masks.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, masks)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    avg_loss = running_loss / len(train_loader)
    loss_history.append(avg_loss)

    # Validation

    model.eval()
    all_probs = []
    all_targets = []

    with torch.no_grad():
        
        for images, masks in val_loader:

            images = images.to(device)
            masks = masks.to(device)
            outputs = model(images)
            preds = torch.sigmoid(outputs)
            all_probs.append(preds.cpu().numpy())
            all_targets.append(masks.cpu().numpy())

    all_probs = np.concatenate(all_probs, axis=0)
    all_targets = np.concatenate(all_targets, axis=0)
    
        
    # Threshold Optimization
    epoch_best_dice = 0.0
    epoch_best_threshold = 0.0
    epoch_false_positives = 0
    epoch_false_negatives = 0

    validation_all_metrics = {
    float(t): {
        "dice": [],
        "false_positives": [],
        "false_negatives": []
    }
    for t in THRESHOLD_RANGE
}

    for t in THRESHOLD_RANGE:
        thresholded_preds = (all_probs > t).astype(np.float32)
        dice_score = global_dice(thresholded_preds, all_targets)
        false_positives = np.sum(thresholded_preds & ~all_targets)/len(all_targets)*100
        false_negatives = np.sum(~thresholded_preds & all_targets)/len(all_targets)*100

        # Store metrics for each threshold
        validation_all_metrics_history.append({
            "threshold": float(t),
            "dice": float(dice_score),
            "false_positives": false_positives,
            "false_negatives": false_negatives
        })

        if dice_score > epoch_best_dice:
            epoch_best_dice = dice_score
            epoch_best_threshold = t
            epoch_false_positives = false_positives
            epoch_false_negatives = false_negatives

    # Store the best metrics for this epoch
    validation_history["best_dice"].append(epoch_best_dice)
    validation_history["best_threshold"].append(epoch_best_threshold)
    validation_history["false_positives"].append(epoch_false_positives)
    validation_history["false_negatives"].append(epoch_false_negatives)

    print(f"Epoch [{epoch + 1}/{MAX_EPOCHS}], Loss: {avg_loss:.4f}, Validation Dice: {epoch_best_dice:.4f}, Best Threshold: {epoch_best_threshold:.2f}, False Positives: {epoch_false_positives:.2f}%, False Negatives: {epoch_false_negatives:.2f}%")

    # Save the best model based on validation Dice score
    if epoch_best_dice > best_global_dice:
        best_global_dice = epoch_best_dice
        best_threshold = epoch_best_threshold
        torch.save(model.state_dict(), os.path.join(OUTPUT_PATH, "best_model.pth"))
        print(f"--- New best model saved with Validation Dice: {best_global_dice:.4f} ---")

# Save the last model
torch.save(model.state_dict(), os.path.join(OUTPUT_PATH, "last_model.pth"))
```

PLOT and ANALYSIS

```python
import pandas as pd

best_model_path = os.path.join(OUTPUT_PATH, "best_model.pth")
last_model_path = os.path.join(OUTPUT_PATH, "last_model.pth")

def load_weights(model, path, device):
    checkpoint = torch.load(path, map_location=device)
    state_dict = checkpoint["model_state_dict"] if isinstance(checkpoint, dict) and "model_state_dict" in checkpoint else checkpoint
    model.load_state_dict(state_dict)

if PLOT_OUTPUT:
    # ----------------------------------------------------
    # 1. Training & Validation Histories
    # ----------------------------------------------------
    # Training Loss
    plt.figure(figsize=(10, 5))
    plt.plot(range(1, MAX_EPOCHS + 1), loss_history, marker='o')
    plt.title('Training Loss History')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.grid(True)
    plt.savefig(os.path.join(OUTPUT_PATH, "loss_history.png"))
    plt.close()

    # Validation Dice
    plt.figure(figsize=(10, 5))
    plt.plot(range(1, MAX_EPOCHS + 1), validation_history["best_dice"], marker='o')
    plt.title('Validation Dice History')
    plt.xlabel('Epoch')
    plt.ylabel('Dice Score')
    plt.grid(True)
    plt.savefig(os.path.join(OUTPUT_PATH, "validation_dice_history.png"))
    plt.close()

    # False Positives / False Negatives
    plt.figure(figsize=(10, 5))
    plt.plot(range(1, MAX_EPOCHS + 1), validation_history["false_positives"], marker='o', label='False Positives')
    plt.plot(range(1, MAX_EPOCHS + 1), validation_history["false_negatives"], marker='o', label='False Negatives')
    plt.title('Validation Error Metrics History')
    plt.xlabel('Epoch')
    plt.ylabel('Count')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(OUTPUT_PATH, "false_positives_negatives_history.png"))
    plt.close()

    # ----------------------------------------------------
    # 2. Last Model Evaluations 
    # ----------------------------------------------------
    print("--- EVALUATING LAST MODEL ---")
    model.eval()

    selected_indices = random.sample(range(len(val_dataset)), min(10, len(val_dataset)))
    
    last_preds_samples = []
    sample_images = []
    sample_targets = []

    # Sample Predictions for Last Model
    with torch.no_grad():
        for idx in selected_indices:
            img, target_mask = val_dataset[idx]
            sample_images.append(img)
            sample_targets.append(target_mask.squeeze().numpy())
            
            pred = model(img.unsqueeze(0).to(device))
            last_preds_samples.append((torch.sigmoid(pred).squeeze().cpu().numpy() > 0.5))

    # Threshold Optimization plot for Last Model
    plt.figure(figsize=(10, 5))
    threshold_dice_scores_last = [
        global_dice((all_probs > t).astype(np.float32), all_targets)
        for t in THRESHOLD_RANGE
    ]
    plt.plot(THRESHOLD_RANGE, threshold_dice_scores_last, marker='o', color='tab:orange')
    plt.title('Threshold Optimization for Last Model')
    plt.xlabel('Threshold')
    plt.ylabel('Dice Score')
    plt.grid(True)
    plt.savefig(os.path.join(OUTPUT_PATH, "last_threshold_optimization.png"))
    plt.close()

    # ----------------------------------------------------
    # 3. Best Model Evaluations
    # ----------------------------------------------------
    print("--- EVALUATING BEST CHECKPOINT ---")

    model.load_state_dict(torch.load(best_model_path, map_location=device))
    model.eval()

    # Regenerate validation probabilities for threshold optimization
    best_probs = []
    with torch.no_grad():
        for images, _ in val_loader:
            outputs = model(images.to(device))
            best_probs.append(torch.sigmoid(outputs).cpu().numpy().astype(np.float32))
    best_probs = np.concatenate(best_probs, axis=0)

    # Sample Predictions for Best Model
    best_preds_samples = []
    with torch.no_grad():
        for img in sample_images:
            pred = model(img.unsqueeze(0).to(device))
            best_preds_samples.append((torch.sigmoid(pred).squeeze().cpu().numpy() > 0.5))

    # Threshold Optimization plot for Best Model
    plt.figure(figsize=(10, 5))
    threshold_dice_scores_best = [
        global_dice((best_probs > t).astype(np.float32), all_targets)
        for t in THRESHOLD_RANGE
    ]
    plt.plot(THRESHOLD_RANGE, threshold_dice_scores_best, marker='o', color='tab:blue')
    plt.title('Threshold Optimization for Best Model')
    plt.xlabel('Threshold')
    plt.ylabel('Dice Score')
    plt.grid(True)
    plt.savefig(os.path.join(OUTPUT_PATH, "best_threshold_optimization.png"))
    plt.close()

    

    # ----------------------------------------------------
    # 4. Side-by-Side Visual Comparison Plot
    # ----------------------------------------------------
    print("--- COMPARISON PLOT ---")
    n_samples = len(selected_indices)
    fig, axes = plt.subplots(n_samples, 3, figsize=(10, 3.5 * n_samples))
    axes = np.atleast_2d(axes)

    for row in range(n_samples):
        # Ground Truth
        axes[row, 0].imshow(sample_targets[row], cmap="gray")
        axes[row, 0].set_title(f"Target — Sample {selected_indices[row]}")
        axes[row, 0].axis("off")

        # Last Model
        axes[row, 1].imshow(last_preds_samples[row], cmap="gray")
        axes[row, 1].set_title("Last Model")
        axes[row, 1].axis("off")

        # Best Model
        axes[row, 2].imshow(best_preds_samples[row], cmap="gray")
        axes[row, 2].set_title("Best Checkpoint")
        axes[row, 2].axis("off")

    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_PATH, "prediction_examples_comparison.png"))
    plt.show()
    plt.close()

# Save the metrics
# Organize metrics by epoch and threshold
metrics_per_epoch = len(THRESHOLD_RANGE)

validation_history["threshold_metrics"] = []

for start in range(0, len(validation_all_metrics_history), metrics_per_epoch):
    epoch_records = validation_all_metrics_history[
        start:start + metrics_per_epoch
    ]

    epoch_metrics = {
        record["threshold"]: {
            "dice_score": record["dice"],
            "false_positives": record["false_positives"],
            "false_negatives": record["false_negatives"],
        }
        for record in epoch_records
    }

    validation_history["threshold_metrics"].append(epoch_metrics)

rows = []

for epoch, train_loss in enumerate(loss_history, start=1):
    epoch_metrics = validation_history["threshold_metrics"][epoch - 1]

    for threshold, metrics in epoch_metrics.items():
        row = {
            "epoch": epoch,
            "average_training_loss": train_loss,
            "threshold": threshold,
        }
        row.update(metrics)
        rows.append(row)

metrics_df = pd.DataFrame(rows)
metrics_df.to_csv(
    os.path.join(OUTPUT_PATH, "training_validation_metrics.csv"),
    index=False
)

print(f"Saved metrics to {os.path.join(OUTPUT_PATH, 'training_validation_metrics.csv')}")
```

TESTING AND SUBMISSION

```python
model.load_state_dict(torch.load(last_model_path, map_location=device))

test_ids = []
test_preds = []

with torch.no_grad():
    for images, image_ids in test_loader:
        images = images.to(device)
        
        outputs = model(images)
        probs = torch.sigmoid(outputs)

        preds = (probs > best_threshold).cpu().numpy()
        
        # Remove the channel dimension so masks are shape (256, 256)
        preds = np.squeeze(preds, axis=1) 
        
        test_preds.append(preds)
        test_ids.extend(image_ids)

test_preds = np.concatenate(test_preds, axis=0)
submission_path = os.path.join(OUTPUT_PATH, "model1_submission.csv")
write_submission(record_ids=test_ids, masks=test_preds, path=submission_path)

print(f"Submission saved successfully to: {submission_path}")
```
